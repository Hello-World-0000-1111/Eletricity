```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: Better plot style
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
df  = pd.read_csv("/content/cleaned_data.csv")
```

```python
df.head(5)
```

```python
# 1. Distribution of Electricity Consumption (Histograph)
plt.figure(figsize=(8,5))
sns.histplot(df['electricity_units_kwh'], bins=30, kde=True)
plt.title("Distribution of Electricity Consumption")
plt.xlabel("Electricity Units (kWh)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(15,5))
plt.plot(df.index, df['electricity_units_kwh'])
plt.title("Daily Electricity Consumption")
plt.xlabel("Date")
plt.ylabel("Electricity Units (kWh)")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='temperature_c',
    y='electricity_units_kwh',
    data=df
)
plt.title("Temperature vs Electricity Consumption")
plt.xlabel("Temperature (°C)")
plt.ylabel("Electricity Units (kWh)")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='humidity_percent',
    y='electricity_units_kwh',
    data=df
)
plt.title("Humidity vs Electricity Consumption")
plt.xlabel("Humidity (%)")
plt.ylabel("Electricity Units (kWh)")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(
    x='household_size',
    y='electricity_units_kwh',
    data=df
)
plt.title("Household Size vs Electricity Consumption")
plt.xlabel("Household Size")
plt.ylabel("Electricity Units (kWh)")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(
    x='city',
    y='electricity_units_kwh',
    data=df
)
plt.title("Electricity Consumption by City")
plt.xlabel("City")
plt.ylabel("Electricity Units (kWh)")
plt.xticks(rotation=45)
plt.show()
```

```python

plt.figure(figsize=(8,6))
sns.boxplot(
    x='income_level',
    y='electricity_units_kwh',
    data=df
)
plt.title("Electricity Consumption by Encoded Income Level")
plt.show()
```

```python
plt.figure(figsize=(10,8))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")
plt.show()
```
